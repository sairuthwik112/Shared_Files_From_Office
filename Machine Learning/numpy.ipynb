```python
# Normalization of array is nothing but the mean is zero and the standard deviation is 1. It is also called as standardization of array. The formula for normalization is (x-mean)/standard deviation. Here x is the value of the array, mean is the mean of the array and standard deviation is the standard deviation of the array.
data=np.array([1,2,3,4,5])
mean=np.mean(data)
standard_deviation=np.std(data)
normalized_data=(data-mean)/standard_deviation
print(normalized_data)
# some other operations on array are median and variance. The median is the middle value of the array and the variance is the measure of how far the values in the array are from the mean. The formula for variance is (sum of (x-mean)^2)/n where x is the value of the array, mean is the mean of the array and n is the number of values in the array.
median=np.median(data)
variance=np.var(data)
print(median)
print(variance)
data[data>3]
```
