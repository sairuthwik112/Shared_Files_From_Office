```python
import pandas as pd
df=pd.read_csv('employees_practice.csv')
df
```

```python
# Handling missing values in a DataFrame is an important step in data preprocessing. Pandas provides several methods to handle missing values, such as dropping rows or columns with missing values, filling missing values with a specific value or using interpolation, and more. Here are some common methods to handle missing values in a DataFrame:
df.isnull()
df.notnull().sum()  # counting the number of non-null values in each column
df.isnull().sum()  # counting the number of null values in each column
df.fillna(0)  # filling the missing values with 0
df['salary_missna']=df['Salary'].fillna(df['Salary'].mean())
df  # filling the missing values in the 'salary' column with the mean of the 'Salary' column
```

```python
df.dtypes
```

```python
df.rename(columns={'Name': 'Employee_Name'},inplace=True)  # Renaming columns in the DataFrame
df
```

```python
df['New Value']=df['Salary'].apply(lambda x: x*2)  # Applying a function to a column in the DataFrame
```

```python
df
```

```python
# Data aggregation is the process of combining multiple data points into a single summary value. In pandas, aggregation can be performed using various methods such as sum(), mean(), median(), min(), max(), count(), etc. Aggregation can be applied to a single column or multiple columns of a DataFrame. Here are some examples of aggregation in pandas:
# Data grouping is the process of grouping data based on one or more columns and applying an aggregation function to each group. In pandas, grouping can be performed using the groupby() method. Here are some examples of grouping in pandas:
# grouped_sum=df.groupby(['Department','City'])['Salary'].sum()  # Grouping the data by 'Department' and 'City' columns and calculating the sum of 'Salary' for each group
# grouped_sum
aggregate_funtion=df.groupby(['City'])['Salary'].agg(['sum','mean','count'])
aggregate_funtion  # Grouping the data by 'City' column and calculating the sum, mean, and count of 'Salary' for each group
```

```python
# Merging and joining data in pandas is the process of combining two or more DataFrames based on a common column or index. Pandas provides several methods for merging and joining data, such as merge(), join(), concat(), etc. Here are some examples of merging and joining data in pandas:
df1=pd.DataFrame({'key':['A','B','C','D'],'value':[1,2,3,4]})
df2=pd.DataFrame({'key':['B','D','E','F'],'value':[5,6,7,8]})
# merged_df=pd.merge(df1,df2,on='key',how='inner')  #
# merged_df=pd.merge(df1,df2,on="key",how="outer")
# merged_df=pd.merge(df1,df2,on="key",how="left") 
merged_df=pd.merge(df1,df2,on="key",how="right") 
# merged_df  # Merging the two DataFrames on the 'key' column using a left join
merged_df  # Merging the two DataFrames on the 'key' column using an inner join  # Merging the two DataFrames on the 'key' column using an inner join
```
