```python
# SQL is a database query language used to create, retrieve, update, and delete data. 
# SQLite is a lightweight relational database engine that uses SQL and stores data in a single file. In Python, the sqlite3 module allows applications to interact with SQLite databases using SQL commands.
import sqlite3

# 1. Connect to the SQLite database
# The database file will be created if it does not exist.
connection = sqlite3.connect("employees.db")

# 2. Create a cursor object
# The cursor executes SQL statements.
cursor = connection.cursor()

# 3. Delete the old table to avoid the missing-column error
# Use this while practising because it also deletes existing employee data.
cursor.execute("DROP TABLE IF EXISTS employees")

# 4. Create the employees table with the correct columns
cursor.execute("""
CREATE TABLE employees (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    department TEXT,
    salary REAL
)
""")

# 5. Insert one row
cursor.execute("""
INSERT INTO employees (name, department, salary)
VALUES (?, ?, ?)
""", ("Sai", "AI Engineering", 50000))

# 6. Insert multiple rows
employee_data = [
    ("Ravi", "Development", 45000),
    ("Priya", "Testing", 40000),
    ("Anil", "Support", 35000)
]

cursor.executemany("""
INSERT INTO employees (name, department, salary)
VALUES (?, ?, ?)
""", employee_data)

# 7. Permanently save the inserted rows
connection.commit()

print("Employee data inserted successfully.")

# 8. Update Sai's salary
cursor.execute("""
UPDATE employees
SET salary = ?
WHERE name = ?
""", (55000, "Sai"))

# 9. Permanently save the update
connection.commit()

print("Employee data updated successfully.")

# 10. Select all rows from the table
cursor.execute("SELECT * FROM employees")

# 11. Retrieve and store all rows
rows = cursor.fetchall()

# 12. Print all rows
print("\nEmployee Data:")

for row in rows:
    print(row)

# 13. Close the database connection
connection.close()
```
